```python
# Create a separate copy for cleaning
cleaned_df = df.copy()

print("Original rows:", len(df))
print("Working copy rows:", len(cleaned_df))
```

```python
income_median = cleaned_df["Income"].median()

print("Income median:", income_median)
```

```python
cleaned_df["Income"] = cleaned_df["Income"].fillna(income_median)
```

```python
print("\nMissing values after cleaning:")
print(cleaned_df.isnull().sum())

print("\nIncome missing values:")
print(cleaned_df["Income"].isnull().sum())
```

```python
print("\n---- Duplicatte Check ----")
print("Duplicate rows:", cleaned_df.duplicated().sum())
```

```python
print("\n----- EDUCATION VALUES BEFORE CLEANING -----")
print(cleaned_df["Education"].unique())

print("\n----- MARITAL STATUS VALUES BEFORE CLEANING -----")
print(cleaned_df["Marital_Status"].unique())
```

```python
print(cleaned_df["Education"].value_counts())
```

```python
print(cleaned_df["Marital_Status"].value_counts())
```

```python
print("\n----- EDUCATION AFTER SPACE CLEANING -----")
print(cleaned_df["Education"].unique())
```

```python
cleaned_df["Marital_Status"] = (
    cleaned_df["Marital_Status"].str.strip()
)
```

```python
print("\n----- MARITAL STATUS AFTER CLEANING -----")
print(cleaned_df["Marital_Status"].unique())
```

```python
print("\n----- DATE VALUES BEFORE CLEANING -----")
print(cleaned_df["Dt_Customer"].head())
```

```python
cleaned_df["Dt_Customer"] = pd.to_datetime(
    cleaned_df["Dt_Customer"],
    dayfirst=True,
    errors="coerce"
)
```

```python
print("\n----- DATE DATA TYPE AFTER CLEANING -----")
print(cleaned_df["Dt_Customer"].dtype)
```

```python
print("\n----- INVALID / MISSING DATES -----")
print(cleaned_df["Dt_Customer"].isna().sum())
```

```python
print("Before cleaning:")
print(cleaned_df.columns)
```

```python
cleaned_df.columns = (
    cleaned_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
print("After cleaning:")
print(cleaned_df.columns)
```

```python
print("Income statistics:")
print(cleaned_df["income"].describe())
```

```python
Q1 = cleaned_df["income"].quantile(0.25)
Q3 = cleaned_df["income"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = cleaned_df[
    (cleaned_df["income"] < lower_bound) |
    (cleaned_df["income"] > upper_bound)
]

print("Number of income outliers:", len(outliers))
```
